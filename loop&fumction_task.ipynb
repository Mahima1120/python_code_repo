```python
#Q1 
# print 1 to 10 using for

for i in range(1,10+1):
    print(i,end=' ')
```

```python
#Q2 
# print 10 to 1 using while

i=10
while i>=1:
    print(i,end=' ')
    i-=1
```

```python
#Q3
#multiplication table

n=int(input("enter a number:"))
i=1
while i<=10:
    print(n,"x",i,"=",n*i)
    i+=1
```

```python
#Q4
#sum of natural numbers

n=int(input("enter a number:"))
sum=0
for i in range(1,n+1):
    sum=sum+i
print("sum = ",sum)
```

```python
#Q5
#pattern 

for i in range(5):
    for j in range(0,i+1):
        print("#",end='')
    print()
    
```

```python
#Q6
#function

def greet(name):
    print("Hello",name)

greet("Mahima")
```

```python
#Q7
#add numbers

def sum(a=0,b=0):
    return a+b

sum=sum(20,40)
print(sum)
```

```python
#Q8
#odd or even

def even(n):
    if n%2==0:
        return True
    else:
        return False
print(even(12))
```

```python
#Q9
#square 

def square(n):
    return n*n
a=square(12)
print(a)
```

```python
#Q10
#large number

def max(a,b):
    if a>b:
        return a
    else:
        return b
 
print(max(10,12))
```
