```python
import numpy as np
```

```python
a=np.array([10,20,30,40])
print(a)
print(a.shape)
print(a.size)
print(a.ndim)
print(a.dtype)
```

```python
a=np.array([[10,20,30],[60,70,80]])
print(a)
print(a.shape)
print(a.size)
print(a.ndim)
print(a.dtype)
```

```python
a=np.array([[[10,20,30],[60,70,80]],[[10,20,30],[40,50,60]]])
print(a)
print(a.shape)
print(a.size)
print(a.ndim)
print(a.dtype)
```

```python
data=np.array([[1,2,3],[4,5,6]],dtype="int")
print(data)
print(data.dtype)
```

```python
data=np.array([[1,2,3],[4,5,6]],dtype="float")
print(data)
print(data.dtype)
```

```python
a=np.full((5,5),[1,2,3,4,5])
print(a)
print(a.dtype)
```

```python
b=np.ones((5,5),dtype="float")
print(b)
print(b.dtype)
```

```python
c=np.zeros((5,5))
print(c)
```

```python
for each in range(-10,11,2):
    print(each,end=" ")
```

```python
print(np.arange(-10,11,2))
```

```python
x=np.linspace(0,100,500)
print(x)
```

```python
print(np.sqrt(20))
print(np.sqrt(-20))
```

```python
print(np.array([10,20])/2)
print(np.array([10,20])/0)
```

```python
a=[1,2,3,4,5]
b=[6,7,8,9,10]
print(a+b)
print(a*b)
```

```python
a=np.array([1,2,3,4,5])
b=np.array([6,7,8,9,10])
print(a+b)
print()
print(a*b)
print()
print(a/b)
print()
print(a-b)
```

```python
a=np.array([1,2,3,4,5])
b=np.array([[2],[4]])
print(a+b)
print()
print(a*b)
print()
print(a/b)
print()
print(a-b)
```

```python
a=["car","bike","van","auto"]
l=[]

for x in a:
    if "a" not in x:
        l.append(x)

print(l)
```
