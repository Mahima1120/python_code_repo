```python
class vehicle:
    x=10
    y=20

    def myfunc(self,a,b):
        print("hello world")
        print(f"this is func variable={a},{b}")

obj=vehicle()

print(obj.x,obj.y)
obj.myfunc("Arun",24)
print()
obj.myfunc("Kavya",28)
print()
obj1=vehicle()
obj1.x="hello"
print(obj1.x,obj1.y)
obj1.myfunc(100,200)
```

```python
class vehicle:
    x=10
    y=20

    def myfunc(self,a,b):
        print("hello world")
        print(f"this is func variable={a},{b}")
        print(f"this is class variable={self.x},{self.y}")

obj=vehicle()
obj.myfunc("hello","world")
```

```python
class car:
    def __init__(self,x,y):
        print("constructor call")
        self.name=x
        self.age=y

obj=car("Priya",20)
print(obj.name,obj.age)
```

```python
class car:
    def __init__(self,x,y):
        print("constructor call")
        self.name=x
        self.age=y

    def __del__(self):
        print("destructor call")

obj=car("Priya",20)
print(obj.name,obj.age)
del obj
```

```python
class India:

    def capital(self):
        print("New Delhi")
    def language(self):
        print("Hindi")

class America:

    def capital(self):
        print("Washington")
    def language(self):
        print("English")

obj_ind=India()
obj_usa=America()

for country in[obj_ind,obj_usa]:
    country.capital()
    country.language()
    print()
```

```python
class xyz:
    x=100
    y=200

obj=xyz()
print(obj.x,obj.y)
obj.x=1000
obj.y=2000
print(obj.x,obj.y)
```

```python
class xyz:

    __x=100
    __y=200

    def getattr(s):
        print(f"this is my class variable={s.__x},{s.__y}")

obj=xyz()
obj.getattr()
obj.x=400
obj.getattr()
```

```python
class car:
    __max=0

    def __init__(self):
        self.__max=200
    def d(self):
        print("driving speed="+str(self.__max))

obj=car()
obj.d()

obj.max=40
obj.d()
```
