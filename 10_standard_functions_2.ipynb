```python
l=[1,2,3,4,5]
def square(x):
    return x*x

ans=list(map(square,l))
print(ans)
```

```python
l=[1,2,3,4,5]
def cube(x):
    return x*x*x

ans=list(map(cube,l))
print(ans)
```

```python
c=["black","blue"]
v=["Bike","Car"]

l=list(zip(c,v))
print(l)
print()
for i in l:
    print(i)
```

```python
c=["black","blue"]
v=["Bike","Car"]

l=list(zip(c,v))
for each in l:
    for i in each:
        print(i,end=" ")
    print()
```

```python
c=["black","blue"]
v=["Bike","Car"]

for c,v in zip(c,v):
    print(c,v)
```

```python
c=["black","blue"]
v=["Bike","Car"]

l=list(zip(c,v))
for each in l:
    print(f"{each[0]}\t{each[1]}")
```

```python
def func():
    yield 100
    yield 200
    yield 300

for i in func():
    print(i)
```

```python
def func():
    print("hello 1")
    yield 100
    print("hello 2")
    yield 200
    print("hello 3")
    yield 300
    print("hello 4")
    yield 400

for i in func():
    print("loop starts")
    print(i)
```
