```python
#Q1
#length of list

mylist=[10,20,30,40,50]
l=len(mylist)
print("length=",l)
```

```python
#Q2
#max and min

num=[2,4,6,8,10]
a=max(num)
b=min(num)
print("max value=",a)
print("min value=",b)
```

```python
#Q3
#sum

num=[1,2,3,4,5,6]
a=sum(num)
print("sum=",a)
```

```python
#Q4
#sort

num=[8,2,4,6,10]
num.sort()
print("ascending=",num)
num.sort(reverse=True)
print("descending=",num)
```

```python
#Q5
#tuple

t=(1,2,3,4,5)
print("tuple=",t)
```

```python
#Q6
#access

mytuple=(1,2,3,4,5)
a=mytuple[0]
b=mytuple[2]
c=mytuple[4]

print("first=",a)
print("middle=",b)
print("last=",c)
```

```python
#Q7
#slice

mytuple=(1,2,3,4,5,6)
first=mytuple[0:3]
last=mytuple[3:6]

print("first=",first)
print("last=",last)
```

```python
#Q8
#remove

a={1,2,3,4,5,6}
a.discard(4)
print("set=",a)
```

```python
#Q9
#pop

a={1,2,3,4,5,6}
b=a.pop()
print("set=",b)
print(a)
```

```python
#Q10
#update

data={
    "name":"preethi",
    "age":40,
    "city":"mumbai"
}

data["age"]=42
print(data)
```

```python
#Q11
#remove

data={
    "name":"preethi",
    "age":40,
    "city":"mumbai"
}

a=data.popitem()
print("removed=",a)
print("updated=",data)
```

```python
#Q12
#remove

data={
    "name":"preethi",
    "age":40,
    "city":"mumbai"
}

data.clear()
print("data=",data)
```

```python
#Q13
#keys()

data={
    "name":"preethi",
    "age":40,
    "city":"mumbai"
}

keys=data.keys()
print("keys=",keys)
```

```python
#Q14
#key-value pair

data={
    "name":"preethi",
    "age":40,
    "city":"mumbai"
}

a=data.items()
print("items=",a)
```

```python
#Q15
#in

data={
    "name":"preethi",
    "age":40,
    "city":"mumbai"
}

if "city" in data:
    print("key exist")
else:
    print("key does not exist")
```
