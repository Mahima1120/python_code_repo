```python
# 30.9.26 task
# remove duplicates
# using function

mylist = [100, "hello", 20.55, "hello", 100, True, False, 100, "Hello", "hai", True]

def remove_duplicates(mylist):
    result = []

    for item in mylist:
        if item not in result:
            result.append(item)

    return result


output = remove_duplicates(mylist)

print(output)
```
