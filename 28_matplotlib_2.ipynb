```python
import matplotlib.pyplot as plt
import numpy as np

fig,((a1,a2),(a3,a4))=plt.subplots(nrows=2,ncols=2,figsize=(10,5))

a1.plot(np.linspace(0,10,50))

a2.scatter([1,2,3],[1,2,3],c='r',marker="*",s=90)

a=['a','b','c','d']
b=[10,20,30,40]
a3.pie(b,labels=a)

a4.plot(np.linspace(3,6,8))

plt.show()
```
