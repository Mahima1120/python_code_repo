```python
import matplotlib.pyplot as plt
```

```python
plt.figure()

plt.plot()
plt.show();
```

```python
import numpy as np

a=np.array([1,2,3])
b=np.array([1,2,3])
c=["r","g","y"]
plt.scatter(a,b,color=c,marker="*",s=150)
plt.show();
```

```python
a=np.array([1,2,3,4,5])
b=np.array([2,4,3,4,2])
c=['red','blue','pink','olive','yellow']

plt.plot(a,b,c='g',marker="*",linestyle="dashed")
plt.scatter(a,b,color=c,marker="*",s=150)
plt.show();
```

```python
a=["A","B","C","D"]
b=["10","20","30","40"]
c=["green","blue","pink","red"]
plt.bar(a,b,color=c,align="edge",edgecolor="black")
plt.show();
```

```python
np.random.seed(4)
a=np.random.normal(50,2,100)
plt.hist(a,color="blue",bins=10,edgecolor="black")
plt.show();
```

```python
a=["A","B","C","D"]
b=[1,2,3,4]
exp=[0.05,0,0,0]
c=["green","blue","pink","red"]
plt.pie(b,labels=a,colors=c,explode=exp)
plt.show();
```

```python
y=["2","4","6","8"]
i=[10,20,30,40]
plt.title("plot")
plt.xlabel("y")
plt.ylabel("i")
plt.plot(y,i,c="g",marker="*")
plt.show();
```

```python
a=[1,2,3,4]
b=[2,4,6,8]
c=[1,2,3,8]
plt.title("stock")
plt.plot(a,label="s1")
plt.plot(b,label="s2")
plt.plot(c,label="s3")
plt.legend()
plt.show();
```

```python
a=plt.figure()
b=a.add_subplot()
b.plot([1,2,3,5],[-1,2,3,4],c="g")
plt.show();
```

```python
a,b=plt.subplots()
b.plot([1,2,3],[1,2,3])
plt.show();
```

```python
car={
    "A":2,
    "B":4,
    "C":6
}
a,b=plt.subplots()
b.bar(car.keys(),car.values());
plt.show();
```

```python
car={
    "A":2,
    "B":4,
    "C":6
}
a,b=plt.subplots()
b.barh(car.keys(),car.values());
plt.show();
```
