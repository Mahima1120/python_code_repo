```python
#Q1

import pandas as pd

data = {
    "EmpID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "Name": ["Arun", "Priya", "Kumar", "Divya", "Rahul", "Meena", "Sathish", "Nila", "Vijay", "Kavya"],
    "Department": ["IT", "HR", "IT", "Finance", "IT", "HR", "Finance", "IT", "Sales", "Sales"],
    "Age": [25, 29, 32, 27, 24, 35, 31, 28, 30, 26],
    "Salary": [35000, 42000, 58000, 39000, 30000, 62000, 51000, 47000, 44000, 36000],
    "Experience": [2, 5, 8, 3, 1, 10, 7, 4, 6, 2],
    "City": ["Chennai", "Madurai", "Chennai", "Coimbatore", "Salem", "Chennai", "Madurai", "Coimbatore", "Chennai", "Salem"]
}

df = pd.DataFrame(data)

print("First 5 Rows:")
print(df.head())

print("\nLast 3 Rows:")
print(df.tail(3))

print("\nNumber of Rows and Columns:")
print(df.shape)

print("\nColumn Data Types:")
print(df.dtypes) 
```

```python
#Q2

result = df[(df["Department"] == "IT") & (df["Salary"] > 40000)]

print(result)
```

```python
#Q3

result = df.sort_values(by="Salary", ascending=False)

print(result[["Name", "Department", "Salary"]])
 
```

```python
#Q4

print("Mean Salary:", df["Salary"].mean())
print("Maximum Salary:", df["Salary"].max())
print("Minimum Salary:", df["Salary"].min())
print("Median Salary:", df["Salary"].median())
print("Average Age:", df["Age"].mean())
```

```python
#Q5

result = df.groupby("Department").agg(
    Number_of_Employees=("EmpID", "count"),
    Average_Salary=("Salary", "mean"),
    Highest_Salary=("Salary", "max")
)

result = result.sort_values(by="Average_Salary", ascending=False)

print(result)
```

```python
#Q6

df["AnnualSalary"] = df["Salary"] * 12

print(df[["Name", "Salary", "AnnualSalary"]])
```

```python
#Q7

df_copy = df.copy()

df_copy.loc[0, "City"] = None

df_copy = pd.concat([df_copy, df_copy.iloc[[0]]], ignore_index=True)

print("Missing Values:")
print(df_copy.isnull().sum())

df_copy["City"] = df_copy["City"].fillna("Unknown")

df_copy = df_copy.drop_duplicates()

print("\nCleaned DataFrame:")
print(df_copy)
```

```python
#Q8

def salary_level(salary):
    if salary >= 50000:
        return "High"
    elif salary >= 40000:
        return "Medium"
    else:
        return "Low"

df["SalaryLevel"] = df["Salary"].apply(salary_level)

print(df[["Name", "Salary", "SalaryLevel"]])
```

```python
#Q9

marks = pd.DataFrame({
    "EmpID": [101, 103, 105, 108, 110],
    "TrainingScore": [85, 90, 78, 88, 95]
})

result = pd.merge(df, marks, on="EmpID", how="left")

print(result[["EmpID", "Name", "TrainingScore"]])
```

```python
#Q10

dept_avg = df.groupby("Department")["Salary"].mean()

print("Department with Highest Average Salary:")
print(dept_avg.idxmax())
print("Average Salary:", dept_avg.max())

avg_salary = df["Salary"].mean()

result = df[df["Salary"] > avg_salary]

print("\nEmployees Above Overall Average Salary:")
print(result[["Name", "Department", "Salary"]])
```
