### **while loop**

```python
r=int(input("enter no of rows:"))
s=input("enter symbol")

i=1
while i<=r:
    j=1
    while j<=i:
        print(s,end='')
        j+=1
    print()
    i+=1
    
```

### **for loop**

```python
for x in range(5):
    for y in range(1,6):
        print('*',end='')
    print()
```
