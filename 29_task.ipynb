```python
import matplotlib.pyplot as plt

df1 = df.groupby('age')['thalach'].mean()
fig, x = plt.subplots(figsize=(8, 6))
x.bar(df1.index, df1.values, color='blue')
x.set(title="Heart rate", xlabel="Age", ylabel="thalach")
plt.show()
```
