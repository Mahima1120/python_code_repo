```python
class Myclass:
    def __init__(self, x):
        self.x = x

    def __add__(self, other):
        return f"The total is {self.x} + {other.x} ="+ str(self.x + other.x)

obj1 = Myclass(30)
obj2 = Myclass(40)
print(obj1 + obj2)
```

```python
class myclass:
    def __init__(self,x):
        self.x = x
    def __sub__(self, other):
        return f"the ans is {self.x} - {other.x} = " + str(self.x-other.x)
    
obj1 = myclass(40)
obj2 = myclass(2)
print(obj1 - obj2)
```

```python
class myclass:
    def __init__(self,x):
        self.x = x
    def __mul__(self, other):
        return f"the ans is {self.x} * {other.x} = " + str(self.x*other.x)
    
obj1 = myclass(40)
obj2 = myclass(2)
print(obj1 * obj2)
```

```python
class Myclass:  
    def __init__(self,x):
        self.x = x
    def __truediv__(self,others):
        return(f"The division of {self.x} and {others.x} is ="  +  str (self.x /others.x))
obj1 = Myclass(30)
obj2 = Myclass(40)
print(obj1 / obj2)
```
